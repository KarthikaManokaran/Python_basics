#### print * pyramid shape

```python
def print_pyramid(n):
    for i in range(1,n+1):
        print(" "*(n-i)+ "*" *(2*i-1))
n=5
print_pyramid(n)
```

variable_scope

In Python, variable scope refers to where a variable can be accessed in a program. It determines the visibility of a variable inside functions, classes, or the entire program.

Python mainly follows the LEGB rule.

1. LEGB Rule in Python
Scope	Meaning	Description
L – Local	Inside a function	Variable created inside a function and used only there
E – Enclosing	Inside nested functions	Variable in the outer function of a nested function
G – Global	Entire program	Variable defined outside all functions
B – Built-in	Python built-in names	Predefined functions like print(), len()

Python checks variables in this order:
Local → Enclosing → Global → Built-in
