#### If else statement syntax:
#syntax:
if(condition):
    Statement 1
    Statement 2
    Statement 3
    Statement 4
    
else:
    Statement 1
    Statement 2
    Statement 3
    
print("program terminated")

```python
age = 17
if(age>18):
    print("you are eligible to vote")

else:
    print("You are not fully developed at the age of 18.")
    
print("Program terminated") # rest of the program 
```

#### nested if

if(condition 1):     # If condition 1 is true, then it goes to condition 2.
    if(condition 2):
        statement 1

#### 4. if - elif - else

syntax

### true condition only works(efficiant & optimization)
if(condition):
    staement 1
elif(condition 2):
    statement 2
elif(condition 3):
    statement 3

```python
### example program:

x=1
if(x==1):
    print("one")
if(x==2):
    print("two")
if(x==3):
    print("three")
if(x==4):
    print("four")
if(x==5):
    print("five")
```

```python
### example program:

x=3
if(x==1):
    print("one")
elif(x==2):
    print("two")
elif(x==3):
    print("three")
elif(x==4):
    print("four")
elif(x==5):
    print("five")
else:
    print("number not found")
print("program terminated")
```
