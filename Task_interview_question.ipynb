```python
## Find the 1st occurance of "n" in string "pythonn"
```

```python
course ="pythonn"
first_occurance =course.find("n")
print(first_occurance)
```

```python
## Find the 2nd occurance of "n" in string "pythonn"
```

```python
course="pythonn"
first_occurance=course.find("n")
print(course.find("n",first_occurance+1))
```

```python
## Find the 2nd occurance of "benx" in LIST 
```

```python
car_list=['bmw','hynudai','benz','benz']
first=car_list.index('benz')
print(car_list.index('benz',first+1))
```
